```python
### **Calculate Recency**
```

### Observation: Some customers (e.g., 13047, 17850) purchased on the very last day of the dataset, giving them a recency score of 0. These are the most recent buyers and represent high engagement.

```python
# Load the raw dataset once at the start
df_raw = pd.read_csv(
    r"C:\Users\Carolina_Muñoz\Desktop\ecommerce-customer-analytics\data\Ecommerce_Sales.csv",
    parse_dates=["InvoiceDate"]   # ensures InvoiceDate is datetime
)

# 1. Work with the raw dataset (with valid dates)
df_ = df_raw.copy()

# 2. Keep only customers with valid IDs
df_clean = df_.dropna(subset=["CustomerID"])   # <-- use df_ here, not df

# 3. Find the most recent purchase date for each customer
last_purchase = df_clean.groupby("CustomerID")["InvoiceDate"].max()

# 4. Define the reference date (latest date in the dataset)
reference_date = df_clean["InvoiceDate"].max()

# 5. Calculate days since last purchase
days_since_last = (reference_date - last_purchase).dt.days

# 6. Put results into a DataFrame
recency_table = days_since_last.reset_index().rename(
    columns={"InvoiceDate":"Recency"}
)

print(recency_table.head())
print(recency_table.shape)

```

```python
###**Calculate Frequency**
```

### Observation: The majority of customers make only one or two purchases, showing that most are occasional buyers. However, a smaller group has very high purchase frequency, which indicates strong loyalty and consistent engagement with the store.

```python
# Frequency = number of unique invoices per customer
frequency = df_clean.groupby("CustomerID")["InvoiceNo"].nunique()

# Convert to DataFrame and rename column
frequency_table = frequency.reset_index().rename(columns={"InvoiceNo":"Frequency"})

print(frequency_table.head())
```

```python
### **Calculate Monetary**
```

### Observation: Customer spending is unevenly distributed. While most customers contribute relatively small amounts, a minority of high‑value customers generate a large share of total revenue. This indicates that the business relies heavily on a small group of big spenders.

```python
df_clean.head()
```

```python
TotalPrice_Sum = df_clean.groupby("CustomerID")["TotalPrice"].sum()
TotalPrice_Sum
```

```python
# Monetary = sum of TotalPrice per customer
monetary = df_clean.groupby("CustomerID")["TotalPrice"].sum().reset_index()

# Rename the column to make it clear
monetary = monetary.rename(columns={"TotalPrice":"Monetary"})

print(monetary.head())
```

```python
### **Combine into RFM table**
```

```python
rfm = recency_table.merge(frequency_table, on="CustomerID") \
                   .merge(monetary, on="CustomerID")

print(rfm.head())
print(rfm.shape)
```

```python
#Create RFM segments
```

### Observation: The segmentation logic correctly identifies high-value customers as "Champions" (e.g., RFM score 555), while others with low frequency and spending but recent activity fall into "Needs Attention." This highlights the importance of not only recent purchases but also consistent frequency and monetary value in defining customer loyalty.


```python
# =====================================================================
# STEP 1: SCORE EACH METRIC
# =====================================================================
def safe_rfm_cut(series, q=5, labels=None, reverse=False):
    if labels is None:
        labels = list(range(1, q+1))
    if reverse:
        labels = labels[::-1]

    if series.nunique() == 1:
        return pd.Series([labels[0]] * len(series), index=series.index)

    try:
        return pd.qcut(series, q=q, labels=labels, duplicates='drop')
    except ValueError:
        return pd.cut(series, bins=q, labels=labels)

# Calculate your 1-5 scores directly into the rfm dataframe
rfm['R_Score'] = safe_rfm_cut(rfm['Recency'], q=5, labels=[1,2,3,4,5], reverse=True)
rfm['F_Score'] = safe_rfm_cut(rfm['Frequency'], q=5, labels=[1,2,3,4,5])
rfm['M_Score'] = safe_rfm_cut(rfm['Monetary'], q=5, labels=[1,2,3,4,5])


# =====================================================================
# STEP 2: COMBINE YOUR SCORES INTO A 3-DIGIT STRING
# =====================================================================
# This line will no longer throw a KeyError because R_Score was just created above
rfm['RFM_Score'] = (
    rfm['R_Score'].astype(str) + 
    rfm['F_Score'].astype(str) + 
    rfm['M_Score'].astype(str)
)


# =====================================================================
# STEP 3: DEFINE THE SEGMENTATION LOGIC
# =====================================================================
def map_rfm_segments(row):
    r = int(row['R_Score'])
    f = int(row['F_Score'])
    m = int(row['M_Score'])
    
    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'
    elif f >= 4 and m >= 4:
        return 'Loyal'
    elif r <= 2 and (f >= 3 or m >= 3):
        return 'At Risk'
    elif r <= 2 and f <= 2 and m <= 2:
        return 'Lost'
    else:
        return 'Needs Attention'


# =====================================================================
# STEP 4: APPLY THE LOGIC AND PREVIEW
# =====================================================================
rfm['Segment'] = rfm.apply(map_rfm_segments, axis=1)

# Save the final table to a fresh CSV file
rfm.to_csv(r"C:\Users\Carolina_Muñoz\Desktop\ecommerce-customer-analytics\data\final_rfm_results.csv", index=False)

print("Process completed successfully! Here is your 3-digit score preview:")
print(rfm[['Recency', 'R_Score', 'Frequency', 'F_Score', 'Monetary', 'M_Score', 'RFM_Score', 'Segment']].head())

```
